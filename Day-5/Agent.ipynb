```python
import os
from dotenv import load_dotenv, find_dotenv


load_dotenv()
```

```python
from langchain_groq import ChatGroq 
from langchain.agents import create_agent
from langchain_core.tools import tool
```

```python
@tool
def add_num(a:int, b:int):
    """  
    
    
    """
    return a + b

def multi(c:int, d:int):
    """
    
    
    """

    return c * d

```

```python
bot = ChatGroq(model="openai/gpt-oss-120b")
```

```python
agent = create_agent(model=bot,
                     tools=[add_num, multi],
                     system_prompt="You are a math teacher, and always use tool for calculation.")
```

```python
response = agent.invoke({"messages":[{"role":"user", "content": " What is 2 + 3 and 3 * 3"}]})
```
