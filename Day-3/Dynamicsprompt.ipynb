```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

bot = ChatGoogleGenerativeAI(model = "gemini-3.7-flash", temperature=1.0)
out = StrOutputParser()
```

```python
def transform_case(result:str):
    return result.upper()
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translator and translates input in {language}"},
    {"role": "user", "content": "{query}"}
])
```

```python
chains = prompts | bot | out | transform_case
```

```python
res = chains.invoke({"language":"Hinglish", "query":"I love CP and DSA"})
print(res)
```
