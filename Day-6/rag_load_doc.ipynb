Text Document Loader

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("speech.txt")
texts = loader.load()

len(texts)
oneDoc= texts[0]
print(oneDoc.page_content)
```

PDF Document Loader

```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("Founder-intern-Resume.pdf")
docs = loader.load()
len(docs)
```

```python
oneDoc = docs[0]
print(oneDoc.page_content)
```

CSV File Loader

```python
from langchain_community.document_loaders import CSVLoader
loader = CSVLoader("speech.csv")
docs = loader.load()
print(len(docs))
```

```python
one_doc = docs[5]
print(one_doc.page_content)
```
