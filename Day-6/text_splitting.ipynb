Text Splitter

```python

texts = "Today, I want to speak from the heart about threeToday, I want to speak from the heart about three pillars that define the future of our nation"
```

```python
from dotenv import load_dotenv
load_dotenv()
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter = RecursiveCharacterTextSplitter(chunk_size = 50, chunk_overlap=20)
chunks  = splitter.split_text(texts)
chunks
```

Document Splitter and Pdf splitter


```python
from dotenv import load_dotenv
load_dotenv()
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.document_loaders import TextLoader , PyPDFLoader
loader = TextLoader("speech.txt")
texts = loader.load()
pdfloader = PyPDFLoader("Founder-Intern-Resume.pdf")
pdf = pdfloader.load()
splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap=100)
chunks_text  = splitter.split_documents(texts)
```
